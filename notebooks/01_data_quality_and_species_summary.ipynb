# A100 Treptower Park – Data quality & species summary

This notebook uses the public-display sample only. It does not contain the full operational coordinate dataset.

```python
from pathlib import Path
import pandas as pd

DATA = Path('../data/tree_inventory_sample.csv')
df = pd.read_csv(DATA)
df.head()
```

```python
summary = {
    'records': len(df),
    'unique_tree_ids': df['tree_id'].nunique(),
    'unique_german_species_labels': df['species_de'].nunique(),
    'unique_botanical_names': df['botanical_name'].nunique(),
    'missing_values': int(df.isna().sum().sum()),
}
summary
```

```python
species_frequency = (
    df.groupby(['species_de', 'botanical_name'])
      .size()
      .sort_values(ascending=False)
      .rename('count')
)
species_frequency.head(10)
```
